```python
import pandas as pd
import numpy as np

# Create dataset
data = {
    "Feature1": [10, 20, 30, 40, 50],
    "Feature2": [100, 200, 300, 400, 500],
    "Feature3": [5, 15, 25, 35, 45]
}

df = pd.DataFrame(data)

print("Original Data:\n")
print(df)

# -----------------------------
# a) Min-Max Normalization
# -----------------------------
df_minmax = (df - df.min()) / (df.max() - df.min())

print("\nMin-Max Normalization:\n")
print(df_minmax)

# -----------------------------
# b) Z-score Normalization
# -----------------------------
df_zscore = (df - df.mean()) / df.std()

print("\nZ-score Normalization:\n")
print(df_zscore)

# -----------------------------
# c) Decimal Scaling
# -----------------------------
def decimal_scaling(column):
    max_val = column.abs().max()
    j = len(str(int(max_val)))
    return column / (10 ** j)

df_decimal = df.apply(decimal_scaling)

print("\nDecimal Scaling Normalization:\n")
print(df_decimal)
```
