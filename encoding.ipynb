```python
import pandas as pd

# Create dataset
data = {
    "Staff": [101, 102, 103, 104, 105],
    "Age": [5, 3, 2, 4, 7],
    "Weight": [20, 8, 0.5, 18, 9],
    "Color": ["Brown", "Black", "Yellow", "White", "Brown"]
}

df = pd.DataFrame(data)

print("Original Dataset:\n")
print(df)

# a) One-hot encoding for Color
color_encoded = pd.get_dummies(df["Color"], prefix="Color")

# b) Combine with original dataset (drop original Color column)
df_final = pd.concat([df.drop("Color", axis=1), color_encoded], axis=1)

# c) Display result
print("\nFinal Dataset after One-Hot Encoding:\n")
print(df_final)
```
