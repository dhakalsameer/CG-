```python
import pandas as pd

# Create dataset
data = {
    "CustomerID": ["C001","C002","C003","C004","C004","C005","C006","C007","C008","C008"],
    "Name": ["John","Alice","BOB","BOBY","Eve","eve","Steve","Ramu","mary","Bob"],
    "Age": [25,34,17,29,29,120,-5,None,220,30],
    "JoinDate": ["12/1/2024","11/15/2023","6/1/2022","6/1/2022","12/5/2024","invalid_date","", "1/1/2024","3/5/2023","3/5/2023"],
    "MonthlyCharges": [29.85,56.95,4000,75.5,75.5,45.99,60,49.99,-30,55],
    "Churn": ["No","Yes","No","No","No","Yes","No",None,"Yes","No"]
}

df = pd.DataFrame(data)

print("Original Dataset:\n")
print(df)

# Identify duplicates based on CustomerID
duplicates = df[df.duplicated(subset="CustomerID", keep=False)]
print("\nDuplicate Rows:\n")
print(duplicates)

# Remove duplicates (keep first occurrence)
df_cleaned = df.drop_duplicates(subset="CustomerID", keep="first")

print("\nDataset After Removing Duplicates:\n")
print(df_cleaned)
```
