```python
# Simple Google Colab Code for Practical Exam

import pandas as pd
import numpy as np

# Step 1: Create dataset
data = {
    "CustomerID": ["C001", "C002", "C003", "C004", "C004", "C005", "C006", "C007", "C008", "C008"],
    "Name": ["John", "Alice", "BOB", "BOBY", "Eve", "eve", "Steve", "Ramu", "mary", "Bob"],
    "Age": [25, 34, 17, 29, 29, 120, -5, np.nan, 220, 30],
    "JoinDate": ["12/1/2024", "11/15/2023", "6/1/2022", "6/1/2022", "12/5/2024",
                 "invalid_date", "", "1/1/2024", "3/5/2023", "3/5/2023"],
    "MonthlyCharges": [29.85, 56.95, 4000, 75.5, 75.5, 45.99, 60, 49.99, -30, 55],
    "Churn": ["No", "Yes", "No", "No", "No", "Yes", "No", np.nan, "Yes", "No"]
}

df = pd.DataFrame(data)

# Display dataset
print("Full Dataset:")
print(df)

# 1. Using isnull()
print("\nUsing isnull():")
print(df.isnull())

# isnull() with any()
print("\nisnull() with any():")
print(df.isnull().any())

# isnull() with sum()
print("\nisnull() with sum():")
print(df.isnull().sum())


# 2. Using isna()
print("\nUsing isna():")
print(df.isna())

# isna() with any()
print("\nisna() with any():")
print(df.isna().any())

# isna() with sum()
print("\nisna() with sum():")
print(df.isna().sum())


# 3. Using notna()
print("\nUsing notna():")
print(df.notna())

# notna() with any()
print("\nnotna() with any():")
print(df.notna().any())

# notna() with sum()
print("\nnotna() with sum():")
print(df.notna().sum())


# 4. Using notnull()
print("\nUsing notnull():")
print(df.notnull())

# notnull() with any()
print("\nnotnull() with any():")
print(df.notnull().any())

# notnull() with sum()
print("\nnotnull() with sum():")
print(df.notnull().sum())
```
