```python
import numpy as np
import pandas as pd
from sklearn.neighbors import KNeighborsClassifier

# Create dataset
data = {
    "BMI": [33.6, 26.6, 23.4, 43.1, 35.3, 35.9, 36.7, 25.7, 23.3, 31],
    "Age": [50, 30, 40, 67, 23, 67, 45, 46, 29, 56],
    "Sugar": [1, 0, 0, 0, 1, 1, 1, 0, 0, 1]
}

df = pd.DataFrame(data)

# Features (X) and Target (y)
X = df[["BMI", "Age"]]
y = df["Sugar"]

# Create KNN model (K=3)
knn = KNeighborsClassifier(n_neighbors=3)

# Train model
knn.fit(X, y)

# New data point
new_point = np.array([[43.6, 40]])  # BMI=43.6, Age=40

# Prediction
prediction = knn.predict(new_point)

print("Prediction (Sugar level):", prediction[0])

if prediction[0] == 1:
    print("Result: Diabetic")
else:
    print("Result: Not Diabetic")
```
